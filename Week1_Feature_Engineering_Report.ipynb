## Overview
The workflow began with standardizing all datetime fields to a consistent datetime format to ensure accuracy and enable reliable time-based operations.

An initial assessment of missing values was then performed across all columns to identify data quality issues and guide subsequent cleaning decisions. Particular attention was given to the Opportunity Start Date and Opportunity End Date fields due to their importance for temporal analysis. Where one of these dates was missing, the absent value was logically estimated using the average opportunity duration calculated from records with complete date information. Records lacking both start and end dates were removed, as no meaningful temporal inference could be made.

In addition to date handling, feature engineering was applied to derive an Age column from the learners’ dates of birth. This transformation enabled age-based analysis while preserving the original birth date information.

Final validation checks confirmed that all critical fields were complete, date ranges were consistent, and the dataset was fully prepared for downstream analysis, visualization, and reporting.
